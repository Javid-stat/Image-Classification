## 2. Look at the data

```python
import matplotlib.pyplot as plt
from torchvision import datasets
from src import config

raw = datasets.FashionMNIST(root=config.DATA_DIR, train=True, download=True)
fig, axes = plt.subplots(2, 8, figsize=(12, 3.4))
for ax, i in zip(axes.flat, range(16)):
    image, label = raw[i]
    ax.imshow(image, cmap="gray_r")
    ax.set_title(config.CLASSES[label], fontsize=8)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
